## CICIoV2024


```python
import pandas as pd
import numpy as np

df = pd.read_parquet("../data/interim/canonical.parquet")

print("=" * 45)
print(" TOTAL ROWS PER ATTACK TYPE")
print("=" * 45)

row_counts = df["attack_type"].value_counts()

for attack, count in row_counts.items():
    print(f"• {attack:<15} : {count} Rows", "(", f"{(count / len(df)) * 100:.2f}", "%)")

print("=" * 45)
print("Total Rows : ", len(df))
print()


id_counts = df.groupby("attack_type")["Arbitration_ID"].nunique()

print("=" * 45)
print(" TOTAL ARBITRATION ID UNIK PER ATTACK TYPE")
print("=" * 45)

for attack, count in id_counts.items():
    print(f"• {attack:<15} : {count} ID Unik")

print("=" * 45)
print()


# ============================================================
# ARBITRATION ID → CLASS
# ============================================================

id_class = (
    df.groupby("Arbitration_ID")["attack_type"]
    .agg(lambda x: ", ".join(sorted(x.unique())))
    .reset_index()
    .rename(columns={"Arbitration_ID": "Arbitration_ID", "attack_type": "Class"})
    .sort_values("Arbitration_ID")
)

id_class.style.hide(axis="index")
```
